<h4>MINI PROJECT CHALLENGE</h4>

<p><b>Coach Timothy is going to make this one slightly harder.</b></p>

<p>Imagine the weather station has three locations:</p>

<pre><code>lagos = [27.5, 29.0, 30.5, 28.0, 31.0]
abuja = [25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]</code></pre>

<p>Your task is to create NumPy arrays for all three locations.</p>

<p><b>Then answer:</b></p>

<br>

<p><b>Challenge 1</b><br>
Which location recorded the highest temperature on each day?</p>

<p><b>Challenge 2</b><br>
Create a calculation that increases every recorded temperature by 1°C.</p>

<p><b>Challenge 3</b><br>
Create a calculation that compares Lagos temperatures with Abuja temperatures.</p>

<p><b>Challenge 4</b><br>
Explain why NumPy becomes increasingly useful as the amount of numerical data grows.</p>

```python
lagos = [27.5, 29.0, 30.5, 28.0, 31.0]
abuja = [25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]

lagos_array = np.array(lagos)
abuja_array = np.array(abuja)
ibadan_array = np.array(ibadan)

# 1. Location with highest temperature each day
max_temperature = np.maximum(np.maximum(lagos_array, abuja_array), ibadan_array)

highest_per_day = np.where(max_temperature == lagos_array, 'Lagos',
                  np.where(max_temperature == abuja_array, 'Abuja', 'Ibadan'))

print(highest_per_day)

# 2. Increasing every recorded temperature by 1
increased_lagos_array = lagos_array + 1
increased_abuja_array = abuja_array + 1
increased_ibadan_array = ibadan_array + 1

print(increased_lagos_array)
print(increased_abuja_array)
print(increased_ibadan_array)

# 3. Compare lagos and abuja tempertures
temperature_difference = lagos_array - abuja_array
print(temperature_difference)

# 4. Numpy becomes increasingly useful as the data volume increased because as the volume increases,
# python list takes a longer time to process numerical data than numpy
```

Project Title: Quick Weather Data Analysis

<!-- Project Introduction -->
This report analyzes daily temperature recordings across three major cities: 
Lagos, Abuja, and Ibadan. The objective is to use NumPy operations to determine
peak daily temperatures, compare temperature variations, and show the 
performance advantages of NumPy over Python lists.

```python
# Import Numpy
import numpy as np

# Dataset
lagos = [27.5, 29.0, 30.5, 28.0, 31.0]
abuja = [25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]

# Convert Data to NumPy Arrays
lagos_array = np.array(lagos)
abuja_array = np.array(abuja)
ibadan_array = np.array(ibadan)
```

```python
# Array Inspection
print("Lagos Array:", "| Data Type:", lagos_array.dtype)
print("Abuja Array:", "| Data Type:", abuja_array.dtype)
print("Ibadan Array:",  "| Data Type:", ibadan_array.dtype)
```

```python
# Numerical Operations
max_temperature = np.maximum(np.maximum(lagos_array, abuja_array), ibadan_array)
highest_per_day = np.where(max_temperature == lagos_array, 'Lagos', 
                  np.where(max_temperature == abuja_array, 'Abuja', 'Ibadan'))

print("Highest Location Per Day:", highest_per_day)

# Increase every recorded temperature by 1°C
increased_lagos = lagos_array + 1
increased_abuja = abuja_array + 1
increased_ibadan = ibadan_array + 1

print("Lagos (+1°C):", increased_lagos)
print("Abuja (+1°C):", increased_abuja)
print("Ibadan (+1°C):", increased_ibadan)
```

```python
# Lagos vs Abuja
temperature_difference = lagos_array - abuja_array

print("Temperature Difference (Lagos - Abuja):", temperature_difference)
```

Findings

1. Lagos recorded the highest temperatures for the first three days. Following that, Abuja was the hottest location for the last three days <br>
2. Lagos recorded a significantly hotter temperature than abuja on the first day
3. Abuja showed the greatest overall temperature increase across the days


Conclusion <br>

Numpy makes multi-location analysis efficient with the element-wise math feature it incoporates.
